```python
%reload_ext autoreload
%autoreload 2
import torch
from utils import attention, audio

import whisper
from datasets import load_from_disk
```

```python
tensors_path = "tensors/"
```

```python
dataset = load_from_disk("../tedlium/test.hf").with_format("torch").select(range(100))
```

```python
model, processor = attention.init()
```

```python
model = whisper.load_model("tiny.en")
```

```python
print(dataset[10])
```

```python
example = dataset[10]["audio"]

"""
Promising examples: 2, 3, 4, 10

Not promising examples: 1
""";
```

```python
snippets = torch.load(f"{tensors_path}/snippets.pt").detach()
```

```python
encoder_limit = int((example.size(0) + snippets.size(1)) / 16000 * 50) + 1
encoder_limit
```

```python
snippets.shape
```

```python
c_attns_tgt = []
for s in snippets.to(example.device):
    _, _, _, _, _, c_attns = attention.inference(torch.cat([s, example]), model, processor, skip_special_tokens=True)
    c_attns_stacked_cutoff = torch.stack([torch.stack(c).squeeze() for c in c_attns[1:]]).permute(1, 2, 0, 3)[:, :, :, :encoder_limit][0, 2]
    c_attns_tgt.append(c_attns_stacked_cutoff)
```

```python
max_seq_len = max(map(lambda x: x.size(0), c_attns_tgt))
```

```python
c_attns_tgt_padded = [torch.nn.functional.pad(c, (0, 0, 0, max_seq_len - c.size(0)), value=-0.05) for c in c_attns_tgt]
```

```python
c_attns_tgt_stacked = torch.stack(c_attns_tgt_padded)
```

```python
c_attns_tgt_stacked.shape
```

```python
test = attention.smoothen(c_attns_tgt_stacked, 10)
test.shape
```

```python
html_vid, anim = attention.plot_attns_iters_anim(test, figsize=(10, 10), vmin=-0.05, vmax=0.050, cmap="magma", interval=50)
```
